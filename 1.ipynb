```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv("csv_collection/data.csv", index_col = 0)

data['X.Time'] = pd.to_datetime(data['X.Time'])
data.rename(columns = {'X.Time' : 'datetime'}, inplace=True)
data.columns
```

```python
columns = [col for col in data.columns if col not in ['X','ind']]
X_input = []
Y_input = []
for col in columns : 
    if col not in 'datetime' : 
        if col.startswith(('FBT', 'WVR','WGT','EQK','WSP','ACC')) : 
            X_input.append(col)
        elif col.startswith(('FBG','DIS','CAC')) : 
            Y_input.append(col)

print(X_input)
print()
print(Y_input)
```

```python
num = 0
feb = data.iloc[[num for num, i in enumerate(data["datetime"]) if str(i).split("-")[1] == "02"],:]
feb = feb.reset_index(drop=True)
```

```python
num = 0
mar = data.iloc[[num for num, i in enumerate(data["datetime"]) if str(i).split("-")[1] == "03"],:]
mar = mar.reset_index(drop=True)
```

```python
X_name = [i for i in data.columns if i.startswith(('FBT','WVR','EQK','ACC','WGT'))]
Y_name = [j for j in data.columns if j.startswith(('FBG','CAC'))]

X_train = feb[[x for x in X_name]]
Y_train = feb[[y for y in Y_name]]

X_test = mar[[x for x in X_name]]
Y_test = mar[[y for y in Y_name]]
```

```python
accuracy_scores = []
clf = RandomForestClassifier(n_estimators=100, max_depth=100,random_state=0)
for num, i in enumerate(Y_train):
    clf.fit(X_train, np.round(Y_train[i]))
    predict = clf.predict(X_test)
    print(accuracy_score(np.round(Y_test[i]), predict))
    accuracy_scores.append(accuracy_score(np.round(Y_test[i]), predict))
    print(f"{i} 완료")
plt.plot(range(len(accuracy_scores)), accuracy_scores)
plt.title("accuracy_scores")
plt.show()
```
