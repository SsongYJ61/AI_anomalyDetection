```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../csv_collection/data.csv",index_col = 0)
df.head()
```

```python
# X, ind 굳이 불필요한 컬럼 제거
df.drop(columns = ['X','ind'], inplace = True)
df
```

```python
## X.time을 datetime으로 변환
df['X.Time'] = pd.to_datetime(df['X.Time'])
df.head()
```

```python
df.rename(columns = {'X.Time' : 'time'}, inplace = True)
df.head()
```

## 1월 시각화

```python
# 시간대별 데이터 개수
# 월별로 먼저 보자
## 1월
month_1 = df.loc[df['time'].dt.month == 1, :]
month_1.shape
```

```python
# 1월의 센서별 값 시계열 그래프


plt.figure(figsize = (12,6))
plt.plot(month_1['time'], month_1.values())
```
